```python
import pandas as pd
import glob
import xml.etree.ElementTree as ET
```

# Extracting

```python
def extract_from_csv(file):
    df = pd.read_csv(file)
    return df
```

```python
def extract_from_json(file):
    df = pd.read_json(file, lines=True)
    return df
```

```python
def extract_from_xml(file):
    df = pd.DataFrame(columns=['name', 'height', 'weight'])
    tree = ET.parse(file)
    root = tree.getroot()
    for person in root:
        name = person.find("name").text
        height = float(person.find("height").text)
        weight = float(person.find("weight").text)
        new_row = pd.DataFrame({"name":[name], "height":[height], "weight":[weight]})
        df = pd.concat([df, new_row], ignore_index=True)
    return df
```

# Transform

```python
def transform(df):
    df['height'] = round(df['height']*0.025,2)
    df['weight'] = round(df['weight']*0.45,2)
    return df
```

# load

```python
def load(df):
    df.to_csv("FinalCSV.csv", index=False)
```

# Running ETL

```python
df = pd.DataFrame(columns=["name", "height", "weight"])
```

```python
for file in glob.glob("*.csv"):
    new_df = extract_from_csv(file)
    df = pd.concat([df,new_df], ignore_index=True)
```
